
# Relatório de Gestão – Análise de Desempenho das Redes Sociais

## 1. Caracterização e validação da base

A base analisada contém 583 publicações da empresa B Braun, sendo cada linha correspondente a um post. As colunas apresentam informações relacionadas à publicação, como rede social, data, tipo de conteúdo e diferentes métricas de desempenho. 

Inicialmente, foram identificados valores nulos em algumas variáveis. Alcance orgânico e alcance pago apresentavam 291 valores nulos cada, enquanto views apresentava 290 valores nulos. As demais métricas não apresentavam valores nulos. Também foi verificado que não havia nenhuma linha duplicada na base. 

A análise das frequências mostrou um grande número de valores iguais a zero em algumas variáveis. Em salvos, por exemplo, foram identificadas 369 publicações com valor zero. No alcance pago, os 292 registros existentes apresentavam valor 0, enquanto aproximadamente metade dos dados era nula. Em views, apenas uma linha apresentava valor diferente de zero. Nas impressões pagas, apenas uma linha apresentava valor diferente de zero.  

Diante da quantidade insuficiente de dados, as variáveis views, impressões pagas, alcance pago e alcance orgânico foram excluídas da análise. A taxa de engajamento foi transformada para o formato numérico e a coluna de data foi convertida para o formato datetime. Após o tratamento, a base passou a apresentar 20 colunas e nenhum valor nulo.  


```python

import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("MetaPosts.csv")

def obj_to_float(x):
    return float(x.replace(' ', '').replace(',', '.'))

df['taxa_de_engajamento'] = df['taxa_de_engajamento'].apply(obj_to_float)

df["data_hora_criacao"] = pd.to_datetime(
    df["data_hora_criacao"],
    format="%d/%m/%Y %H:%M:%S"
)

df = df.drop(columns=['views', 'impressoes_pagas', 'alcance_pago', 'alcance_organico'])
```

## 4. Análise de outliers

As variáveis analisadas apresentam muitos outliers, principalmente acima do limite superior, indicando assimetria positiva e alta dispersão.

Por esse motivo, a mediana é considerada mais adequada que a média para representar os valores centrais. Os outliers não devem ser removidos automaticamente, pois podem representar desempenhos excepcionalmente altos.

A comparação entre as médias com e sem outliers permite observar o quanto os valores extremos influenciam os resultados. Quando a média diminui significativamente após a remoção dos outliers, isso indica que alguns valores muito elevados estavam aumentando a média original. Dessa forma, a média sem outliers pode representar melhor o comportamento típico das publicações.


**Gráfico boxplot do alcance total**

```python
df['alcance_total'].plot.box()
plt.title("Distribuição do Alcance Total")
plt.ylabel("Alcance Total")
plt.tight_layout()
plt.show()
```

**Quantidade de outliers por métrica**


```python
outliers = df.select_dtypes(include='number').columns
for coluna in outliers:
    q1 = df[coluna].quantile(0.25)
    q3 = df[coluna].quantile(0.75)
    iqr = q3 - q1
    limite_inferior = q1 - 1.5 * iqr
    limite_superior = q3 + 1.5 * iqr
    quantidade = ((df[coluna] < limite_inferior) | (df[coluna] > limite_superior)).sum()
    print(f"{coluna}: {quantidade} outliers")
```


## 8. Definição das metas para 2027

As metas de 2027 foram estabelecidas utilizando as taxas de variação dos seis meses em comum entre 2025 e 2026. Foram analisadas medidas como média, mediana, quartis e desvio padrão.

A mediana foi escolhida como principal referência porque algumas métricas apresentaram variações muito elevadas, principalmente cliques, salvos e compartilhamentos. A taxa mediana de cada métrica foi aplicada aos resultados de 2026 para projetar as metas de 2027.

Os outliers também foram analisados pelo método do Intervalo Interquartil (IQR). A taxa de engajamento foi analisada separadamente por ser uma métrica percentual, utilizando também a mediana como referência. 

### Metas projetadas para 2027:


**Média e mediana das taxas de variação por métricas nos meses em comum de 2025 e 2026**

```python
new_a25 = df[df['ano'] == 2025]
new_mes_comum = new_a25[new_a25['mes'].isin([1,2,3,4,5,6])]
new_metricas_a25 = new_mes_comum.groupby('mes')[[
    "alcance_total", "curtidas_e_reacoes", "comentarios",
    "compartilhamentos", "cliques", "salvos",
    "impressoes_organicas", "impressoes_totais", "envolvimento"]].sum()

new_a26 = df[df['ano'] == 2026]
new_mes_comum = new_a26[new_a26['mes'].isin([1,2,3,4,5,6])]
new_metricas_a26 = new_mes_comum.groupby('mes')[[
    "alcance_total", "curtidas_e_reacoes", "comentarios",
    "compartilhamentos", "cliques", "salvos",
    "impressoes_organicas", "impressoes_totais", "envolvimento"]].sum()

taxa_de_variacao = ((new_metricas_a26 - new_metricas_a25) / new_metricas_a25)

resumo_taxas = pd.DataFrame({
    "Média": taxa_de_variacao.mean(),
    "Mediana": taxa_de_variacao.median(),
}).round(2)
resumo_taxas
```

```python
new_a25 = df[df['ano'] == 2025]
new_mes_comum = new_a25[new_a25['mes'].isin([1,2,3,4,5,6])]
taxa_engajamento25 = new_mes_comum.groupby('mes')[['taxa_de_engajamento']].mean()
 
new_a26 = df[df['ano'] == 2026]
new_mes_comum = new_a26[new_a26['mes'].isin([1,2,3,4,5,6])]
taxa_engajamento26 = new_mes_comum.groupby('mes')[['taxa_de_engajamento']].mean()
 
comparativo_engajamento = taxa_engajamento25.join(
    taxa_engajamento26, lsuffix='_2025', rsuffix='_2026'
)
taxa_variacao_engajamento = (
    (taxa_engajamento26 - taxa_engajamento25) / taxa_engajamento25
)
taxa_variacao_engajamento

 
```
