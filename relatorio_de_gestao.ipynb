# RELATÓRIO DE GESTÃO — ANÁLISE DE MÍDIAS SOCIAIS

## 1. Introdução

Este relatório apresenta a análise dos dados de publicações das redes sociais Instagram e Facebook, utilizando uma base com 583 registros e 24 variáveis inicialmente disponíveis.

Foram analisadas métricas relacionadas ao alcance, curtidas e reações, comentários, compartilhamentos, cliques, salvos, impressões, envolvimento e taxa de engajamento.

Para a comparação entre 2025 e 2026, foram considerados os meses em comum entre os dois períodos. Como os dados de 2026 não estavam completos para todos os meses, a análise comparativa final utilizou os meses de janeiro a junho.

---

## 2. Validação e tratamento da base

Inicialmente, foi realizada uma análise da estrutura da base. Foram identificados 583 registros e 24 colunas.

Algumas variáveis apresentaram quantidade significativa de valores nulos. As variáveis `views`, `alcance_pago` e `impressoes_pagas` apresentaram dados insuficientes para utilização na análise. Em `views`, apenas uma linha apresentava valor diferente de zero. Em `alcance_pago`, metade dos dados era nula e a outra metade apresentava valor igual a 0. Já em `impressoes_pagas`, apenas uma linha apresentava valor diferente de zero.

Também foi verificado que não existiam linhas duplicadas na base.

A taxa de engajamento inicialmente estava armazenada como texto e foi convertida para o formato numérico. A variável de data também foi convertida para o formato `datetime`.

As colunas com quantidade insuficiente de dados foram excluídas. Para os valores nulos de `alcance_organico`, foi utilizada a mediana para preenchimento, considerando a presença de outliers.

Após o tratamento, a base passou a não apresentar valores nulos nas variáveis utilizadas na análise.

---

## 3. Análise estatística das métricas

### 3.1 Impressões Totais

As impressões totais apresentam alto desvio padrão, indicando alta dispersão dos dados.

A média é de **3.303,78**, enquanto a mediana é de **1.369**. Essa diferença indica uma assimetria positiva, mostrando que alguns valores muito altos elevam a média.

Os quartis mostram que:

* 25% dos dados estão abaixo de **310**;
* 50% estão abaixo de **1.369**;
* 75% estão abaixo de **3.732**.

Dessa forma, metade dos dados está concentrada entre **310 e 3.732 impressões**.

O valor máximo é de **145.970**, muito superior ao terceiro quartil, à média e à mediana, indicando a presença de possíveis outliers.

O intervalo interquartil (IQR) é de **3.422**. Considerando o critério de 1,5 × IQR, valores acima de aproximadamente **8.865** podem ser considerados possíveis outliers.

### 3.2 Alcance Total

O alcance total também apresenta alto desvio padrão, indicando alta dispersão dos dados.

A média é de **1.600,88**, enquanto a mediana é de **841**. A média é aproximadamente o dobro da mediana, indicando uma assimetria positiva, na qual alguns valores altos elevam a média.

Os quartis mostram que:

* 25% dos dados estão abaixo de **167,5**;
* 50% estão abaixo de **841**;
* 75% estão abaixo de **2.068**.

Assim, metade dos valores está concentrada entre **167,5 e 2.068** de alcance.

O valor máximo é de **91.421**, muito superior ao terceiro quartil, à média e à mediana, indicando a presença de possíveis outliers.

O intervalo interquartil é de **1.900,5**. Pelo critério de 1,5 × IQR, valores acima de aproximadamente **4.919** podem ser considerados possíveis outliers.

### 3.3 Curtidas e Reações

A quantidade de curtidas e reações apresenta alto desvio padrão, indicando que os valores variam consideravelmente entre as publicações.

A média é de **65,96**, enquanto a mediana é de **30**. A média é mais que o dobro da mediana, indicando uma assimetria positiva causada principalmente pela presença de valores elevados.

Os quartis mostram que:

* 25% dos dados estão abaixo de **8**;
* 50% estão abaixo de **30**;
* 75% estão abaixo de **72,5**.

Assim, metade dos valores está concentrada entre **8 e 72,5 curtidas e reações**.

O valor máximo é de **1.020**, muito superior ao terceiro quartil, à média e à mediana, indicando a presença de possíveis outliers.

O intervalo interquartil é de **64,5**. Pelo critério de 1,5 × IQR, valores acima de aproximadamente **169,25** podem ser considerados possíveis outliers.

## 4. Comparação entre 2025 e 2026

Para realizar a comparação entre os anos, foram utilizados os meses em comum entre 2025 e 2026. Inicialmente, foram considerados os meses de **janeiro a julho**, pois esses eram os meses disponíveis nos dois períodos.

A análise foi realizada por meio do agrupamento dos dados por ano, mês e rede social, utilizando a soma das métricas.

O **Instagram apresentou números maiores de alcance, impressões, curtidas e reações na maioria dos meses**.

O **mês 5** apresentou os maiores números de alcance entre os meses analisados por meio do Instagram. Considerando os dois anos, o mês 5 apresentou quase o dobro de alcance em comparação com o segundo mês com maior alcance. Também foi o mês com maior número de impressões, curtidas e reações.

Na soma dos dois anos, o mês 5 apresentou os melhores números para alcance total e curtidas e reações. Também foi observada uma grande queda no mês 6 em alcance e curtidas e reações.

O número de curtidas permaneceu em um nível mais baixo durante os meses analisados.

---

## 5. Comparação mensal entre 2025 e 2026

Na comparação dos anos, foi observado que:

* **Alcance Total:** 2026 foi superior a 2025 nos meses 3, 4, 5 e 6. 2025 foi superior nos meses 1, 2 e 7.
* **Curtidas e Reações:** 2026 foi superior a 2025 nos meses 2, 3, 4, 5 e 6. 2025 foi superior nos meses 1 e 7.
* **Impressões Totais:** 2026 foi superior a 2025 nos meses 3, 4, 5 e 6. 2025 foi superior nos meses 1, 2 e 7.

O início de 2025 apresentou números superiores ao início de 2026. Entretanto, nos meses seguintes, 2026 apresentou uma grande melhora em relação a 2025.

Também foi observada uma grande queda do mês 5 para o mês 6 em 2026.

O mês 7 de 2026 apresentou números muito abaixo de 2025. Entretanto, esse mês possuía dados incompletos, pois os dados disponíveis não abrangiam todo o mês. Por esse motivo, o mês 7 foi desconsiderado na análise final.

De forma geral, 2025 apresentou altos e baixos durante o período analisado, porém manteve uma linha sem meses de crescimento ou queda extremos.

---

## 6. Taxa de variação

Para realizar a projeção, foram considerados os seis meses em comum entre 2025 e 2026, de janeiro a junho.

Inicialmente, foi calculada a diferença absoluta entre os valores dos dois anos. Entretanto, esse método apresentou resultados negativos em alguns casos e, por isso, foi utilizada a taxa de variação.

A taxa de variação foi calculada pela razão entre a diferença dos valores de 2026 e 2025 e o valor de 2025:

**Taxa de variação = (Valor de 2026 − Valor de 2025) / Valor de 2025**

As taxas de variação obtidas para os seis meses foram:

| Mês       | Alcance Total | Curtidas e Reações | Impressões Totais |
| --------- | ------------: | -----------------: | ----------------: |
| Janeiro   |          -57% |               -74% |              -70% |
| Fevereiro |           -8% |               +26% |               -9% |
| Março     |          +27% |               +27% |              +17% |
| Abril     |         +198% |              +163% |             +225% |
| Maio      |          +79% |               +63% |              +67% |
| Junho     |          +17% |               +36% |              +16% |

A média da taxa de variação dos seis meses foi:

| Métrica            | Média da taxa de variação |
| ------------------ | ------------------------: |
| Alcance Total      |                   **43%** |
| Curtidas e Reações |                   **40%** |
| Impressões Totais  |                   **41%** |

---

## 7. Projeção para 2027

Para realizar a projeção, a taxa de variação foi aplicada aos valores de 2026. Dessa forma, foi simulado o comportamento dos seis meses correspondentes de 2027.

A fórmula utilizada foi:

**Valor projetado = Valor de 2026 × (1 + Taxa de variação)**

Os valores projetados foram:

| Mês       | Alcance Total | Curtidas e Reações | Impressões Totais |
| --------- | ------------: | -----------------: | ----------------: |
| Janeiro   |      4.728,53 |             110,29 |          4.910,27 |

Também foi calculado o intervalo de confiança com o objetivo de estimar a faixa na qual os valores projetados podem se situar, considerando a variabilidade observada nos dados.

Os outliers foram mantidos na análise devido à possibilidade de representarem publicações que viralizaram.

---

## 8. Conclusão

A análise dos dados permitiu comparar o desempenho das métricas entre 2025 e 2026 utilizando os meses em comum.

Foi identificado que o Instagram apresentou números maiores de alcance, impressões, curtidas e reações na maioria dos meses. O mês de maio apresentou os melhores resultados nas métricas analisadas, enquanto também foi observada uma queda significativa no mês de junho.

Na comparação entre os anos, 2026 apresentou resultados superiores em grande parte dos meses analisados, principalmente entre março e junho. O mês de julho de 2026 foi desconsiderado por apresentar dados incompletos.

A partir dos seis meses em comum, foi calculada a taxa de variação entre 2025 e 2026 e aplicada aos valores de 2026 para realizar a projeção dos seis meses correspondentes de 2027.

As médias das taxas de variação foram de **43% para Alcance Total, 40% para Curtidas e Reações e 41% para Impressões Totais**.

Os valores projetados foram apresentados para os seis meses analisados, juntamente com o cálculo do intervalo de confiança.
