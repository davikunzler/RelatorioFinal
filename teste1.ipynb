Meses em comum entre 2025 e 2026(meses 1,2,3,4,5,6,7)

```python

a25_a26 = df[df['ano'].isin([2025, 2026])]
mes_comum = a25_a26[a25_a26['mes'].isin([1,2,3,4,5,6,7])]
metricas_mes = mes_comum.groupby(['mes', "ano"])[["alcance_organico","alcance_total", "curtidas_e_reacoes", "comentarios","compartilhamentos", "cliques","salvos","views","impressoes_totais", "impressoes_organicas", "impressoes_pagas", "envolvimento"]].sum()
metricas_mes
```

um ano de cada vez

```python
a25 = df[df['ano'] == 2025]
mes_comum = a25[a25['mes'].isin([1,2,3,4,5,6,7])]
metricas_a25 = mes_comum.groupby('mes')[["alcance_organico","alcance_total", "curtidas_e_reacoes", "comentarios","compartilhamentos", "cliques","salvos","views","impressoes_totais", "impressoes_organicas", "impressoes_pagas", "envolvimento"]].sum()
metricas_a25
```

```python
a26 = df[df['ano'] == 2026]
mes_comum = a26[a26['mes'].isin([1,2,3,4,5,6,7])]
metricas_a26 = mes_comum.groupby('mes')[["alcance_organico","alcance_total", "curtidas_e_reacoes", "comentarios","compartilhamentos", "cliques","salvos","views","impressoes_totais", "impressoes_organicas", "impressoes_pagas", "envolvimento"]].sum()
metricas_a26
```

Taxa de varição das métricas nos meses em comum

```python
taxa_de_variacao = ((metricas_a26 - metricas_a25) / metricas_a25 * 100)
taxa_de_variacao.round(2)
```

Média da taxa de variação entre os meses em comum

```python
media_taxa_de_variacao = taxa_de_variacao.mean().round(2).astype(str) + " %"
media_taxa_de_variacao
```

Tranformar a taxa de engajamento para fazer o calcúlo

```python
df["taxa_de_engajamento"] = (
    df["taxa_de_engajamento"]
    .str.replace(",", ".")
    .astype(float)
)
```

```python
taxa_2025 = (
    a25[a25["mes"].isin([1,2,3,4,5,6,7])]
    .groupby("mes")["taxa_de_engajamento"]
    .mean()
)*100
taxa_2025
```

```python
taxa_2026 = (
    a26[a26["mes"].isin([1,2,3,4,5,6,7])]
    .groupby("mes")["taxa_de_engajamento"]
    .mean()
)*100
taxa_2026
```

Taxa de variação da taxa de engajamento

```python
crescimento_taxa = (
    (taxa_2026 - taxa_2025)
    / taxa_2025
)*100
crescimento_taxa.round(2)
```

Média da taxa de variação da taxa de engajamento

```python
taxa_previsao = crescimento_taxa.mean().round(2)
taxa_previsao.astype(str) + " %"
```
